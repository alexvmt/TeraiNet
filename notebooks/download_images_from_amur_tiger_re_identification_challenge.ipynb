## Setup

```
import os
import sys

from google.colab import drive

drive.mount("/content/drive")

# add folder to sys.path so that utilities can be imported
project_dir = "drive/MyDrive/TeraiNet"
sys.path.append(project_dir)
from utilities import load_config
```

```
# load config and set variables and parameters
config_path = os.path.join(project_dir, "config.yaml")
config = load_config(config_path)

scripts_dir = os.path.join(project_dir, config["sampling_downloading"]["scripts_dir"])
prep_amur_tiger_images_script = os.path.join(
    scripts_dir, config["scripts"]["prep_amur_tiger_images_script"]
)

images_dir = os.path.join(project_dir, config["sampling_downloading"]["images_dir"])
```

```
# for a fresh start, remove images dir
remove_images_dir = True
if remove_images_dir:
    !rm -rf "$images_dir"
    !mkdir -p "$images_dir"
```

## Download images

#### Amur tiger re-identification challenge
These images are a substitute for "real" camera trap images because openly available camera trap images of tigers are scarce in general, let alone Bengal tigers.

```
class_name = "tiger"
class_number = config["classes"][class_name]
```

```
# download and unpack train images
subset = "train"
url = f"https://storage.googleapis.com/public-datasets-lila/cvwc2019/{subset}/atrw_detection_{subset}.tar.gz"
file = f"atrw_detection_{subset}.tar.gz"
!wget -O "$file" "$url"
!tar -xzf "$file"
```

```
# rename and move train images
source_dir = "trainval"
!echo "Files in source dir:" $(ls "$source_dir" | wc -l)
!time bash "$prep_amur_tiger_images_script" "$source_dir" "$class_number" "$subset" "$images_dir"
!echo "Files in target dir:" $(ls "$images_dir/"class_"$class_number"_"$subset"* | wc -l)
```

```
# download and unpack test images
subset = "test"
url = f"https://storage.googleapis.com/public-datasets-lila/cvwc2019/{subset}/atrw_detection_{subset}.tar.gz"
file = f"atrw_detection_{subset}.tar.gz"
!wget -O "$file" "$url"
!tar -xzf "$file"
```

```
# rename and move test images
source_dir = "test"
!echo "Files in source dir:" $(ls "$source_dir" | wc -l)
!time bash "$prep_amur_tiger_images_script" "$source_dir" "$class_number" "$subset" "$images_dir"
!echo "Files in target dir:" $(ls "$images_dir/"class_"$class_number"_"$subset"* | wc -l)
```
