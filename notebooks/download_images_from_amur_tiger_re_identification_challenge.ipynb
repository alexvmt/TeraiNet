## Download images

#### Amur tiger re-identification challenge
These images are a substitute for "real" camera trap images because openly available camera trap images of tigers are scarce in general, let alone Bengal tigers.

```python
class_name = 'tiger'
class_number = config['classes'][class_name]
```

```python
# download and unpack train images
subset = 'train'
url = f'https://storage.googleapis.com/public-datasets-lila/cvwc2019/{subset}/atrw_detection_{subset}.tar.gz'
file = f'atrw_detection_{subset}.tar.gz'
!wget -O "$file" "$url"
!tar -xzf "$file"
```

```python
# rename and move train images
source_dir = 'trainval'
!echo "Files in source dir:" $(ls "$source_dir" | wc -l)
!time bash "$prep_amur_tiger_images_script" "$source_dir" "$class_number" "$subset" "$images_dir"
!echo "Files in target dir:" $(ls "$images_dir/"class_"$class_number"_"$subset"* | wc -l)
```

```python
# download and unpack test images
subset = 'test'
url = f'https://storage.googleapis.com/public-datasets-lila/cvwc2019/{subset}/atrw_detection_{subset}.tar.gz'
file = f'atrw_detection_{subset}.tar.gz'
!wget -O "$file" "$url"
!tar -xzf "$file"
```

```python
# rename and move test images
source_dir = 'test'
!echo "Files in source dir:" $(ls "$source_dir" | wc -l)
!time bash "$prep_amur_tiger_images_script" "$source_dir" "$class_number" "$subset" "$images_dir"
!echo "Files in target dir:" $(ls "$images_dir/"class_"$class_number"_"$subset"* | wc -l)
```
