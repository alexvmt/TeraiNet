## Download images

1. Specify `class_name` in the cell below.
2. Run the two cells below and wait until all images of the specified class have been downloaded.
3. Download `images` directory from Drive.
4. Restart notebook and start over with next class.

Repeat steps until images for all classes have been downloaded successfully.
This process is somewhat tedious but free Drive has only 15 GB space and downloading all images at once would be too much. Remember to regularily empty the recycle bin to really free up space.

Once the process is complete, use `unzip_and_copy_lila_bc_images.sh` to unzip and copy all downloaded images to one joint `images` directory. Run the script in the directory where your downloaded compressed files are located (e. g. copy it there). Make sure there are no other zip-files in that directory. Note that Drive might split your `images` directory into several files during compression right before the download starts so that there might be more compressed files to process than classes.

Use `tar -czf images.tar.gz images/*` to compress the created joint `images` directory that can then be uploaded to Kaggle as a dataset.
