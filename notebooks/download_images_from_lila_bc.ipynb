```python
def load_config(config_path):
    with open(config_path, 'r') as file:
        config = yaml.safe_load(file)
    return config
```

```python
project_dir = 'drive/MyDrive/TeraiNet'
config_path = os.path.join(project_dir, 'config.yaml')
config = load_config(config_path)

scripts_dir = os.path.join(project_dir, config['directories']['scripts_dir'])
download_images_script = os.path.join(scripts_dir, config['scripts']['download_images_script'])

samples_dir = os.path.join(project_dir, config['directories']['samples_dir'])

images_dir = os.path.join(project_dir, config['directories']['images_dir'])
```
