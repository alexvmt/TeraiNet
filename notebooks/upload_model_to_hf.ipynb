```python
from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
hf_token = user_secrets.get_secret("HF_TOKEN")
```

```python
REPO_ID = "alexvmt/TeraiNet"
```

```python
from pathlib import Path

model_path = Path("/kaggle/input/notebooks/alexvmt/train-and-evaluate-terainet/model.keras")

print(model_path.exists())
print(model_path)
```

```python
from huggingface_hub import HfApi

api = HfApi(token=hf_token)

api.upload_file(
    path_or_fileobj=str(model_path),
    path_in_repo="model.keras",
    repo_id=REPO_ID,
    repo_type="model",
)
```

```python
additional_files = [
    "class_list.yaml",
    "preparation_summary.yaml",
    "training_config.yaml",
]

for filename in additional_files:
    local_path = Path("/kaggle/input/notebooks/alexvmt/train-and-evaluate-terainet") / filename

    if local_path.exists():
        api.upload_file(
            path_or_fileobj=str(local_path),
            path_in_repo=filename,
            repo_id=REPO_ID,
            repo_type="model",
        )
```
